```python
from sklearn.neighbors import KNeighborsClassifier

# Training data
# [Study Hours, Attendance]
X = [
    [2, 60],
    [3, 65],
    [4, 70],
    [5, 75],
    [6, 80],
    [7, 85],
    [8, 90],
    [9, 95]
]

# 0 = Fail, 1 = Pass
y = [0, 0, 0, 1, 1, 1, 1, 1]

# Create KNN model with K = 3
model = KNeighborsClassifier(n_neighbors=3)

# Train the model
model.fit(X, y)

# New student's data
new_student = [[6, 78]]

# Make prediction
prediction = model.predict(new_student)

# Display result
if prediction[0] == 1:
    print("Prediction: Student will PASS")
else:
    print("Prediction: Student will FAIL")
```

classify fruits based on weight and size

```python
from sklearn.neighbors import KNeighborsClassifier

# Training data
# [weight, size]
X = [
    [150, 7],
    [160, 7],
    [170, 8],
    [180, 8],
    [250, 10],
    [260, 10],
    [270, 11],
    [280, 11]
]

# Labels
# 0 = Apple, 1 = Orange
y = [0, 0, 0, 0, 1, 1, 1, 1]

# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)

# Train the model
model.fit(X, y)

# New fruit
new_fruit = [[175, 8]]

# Predict
prediction = model.predict(new_fruit)

if prediction[0] == 0:
    print("Predicted Fruit: Apple")
else:
    print("Predicted Fruit: Orange")
```
