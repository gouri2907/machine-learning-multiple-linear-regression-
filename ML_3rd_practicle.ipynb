```python
Data={"Area":[800,900,1000,1100,1250,1500,1500,1700,1900,1900,2000,2500],
      "BHK":[1,1,1,2,2,2,3,3,3,4,4,4],
      "price":[4000000,4100000,4500000,4750000,5200000,5700000,6000000,6200000,6500000,6750000,6800000,7000000]}
```

```python
print(Data)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
df=pd.DataFrame(Data)
```

```python
print(df)
```

```python
plt.scatter(df["Area"],df["BHK"],df["price"])
plt.scatter(df["BHK"],df["price"])
plt.scatter(df["price"],df["Area"])
```

```python
df.corr()
```

```python
sns.pairplot(df)
```

feature selection

```python
X=df[["Area","BHK"]]
```

```python
y=df["price"]
```

train test split

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.14,random_state=42)
```

```python
X_train
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

view model parameters

```python
print("intersept")
model.intercept_
model.coef_
```

make prediction

```python
y_pred=model.predict(X_test)
print(y_pred)
```

```python
print("actual value",y_test.values)
```

```python
mse=mean_squared_error(y_test,y_pred)
print(mse)
```

```python
print(r2_score(y_test,y_pred))
```
